```python
from google import genai
from getpass import getpass #getpass helps us in passing our API key secretly
```

```python
#entering API Key securely
api_key=getpass("Enter your Gemini API Key: ")
```

```python
#creating Gemini Client and generating a response
client=genai.Client(api_key=api_key)
```

```python
user_message="Explain the meaning of this term called EBITDA in simple language"
```

```python
response=client.models.generate_content(
    model="gemini-3.6-flash",
    contents=user_message #based on the user_message, a response will be generated using gemini-3.6-flash model
)
```

```python
print(response.text)
```

Messages are of two types:

- User Message / Human Message - this is the query which comes from the user
- System Message - this allows us to manage the behaviour of our LLM

```python
#Creating helper function - so that we dont have to repeat the code where we do API call every single time
from google import genai
from google.genai import types

client=genai.Client(api_key=api_key)

#creating the function now
def ask_gemini(user_message,system_message=None):
  response=client.models.generate_content(
      model="gemini-3.6-flash",
      contents=user_message)

  return response.text
```

```python
result=ask_gemini(
    user_message="Explain the difference between AI and GenAI"
)
```

```python
print(result)
```

- Zero Shot Prompting - ask the model to perform a task without giving examples

```python
prompt="""
Classify the following customer review as - 'Positive', 'Negative' or 'Neutral'

Review - "The product arrived on time and works perfectly"
"""
```

```python
response=ask_gemini(prompt)
print(response)
```

- Few-Shot Prompting

```python
prompt="""
You are classifying banking customer queries and the categories are :
- Loan
- Credit Card
- Account
- Transaction
- Others

Examples

Query: "I want to know the interest rate for a home loan"
Category: Loan

Query: "How do I open a new credit card?"
Category: Credit Card

Query: "My credit card payment is showing declined"
Category: Credit Card

Query: "I cannot see my latest transaction history"
Category: Transaction

Now classify -

Query: "I want to increase my credit card limit"
Category:

Query: "How do I open a new bank account"
Category
"""
```

```python
response=ask_gemini(prompt)
```

```python
print(response)
```

- Chain of Thought

```python
prompt="""
You are a banking credit analyst and you are supposed to consider this company data:

- Revenue = 100 Cr
- EBITDA = 20 Cr
- Debt = 60 Cr
- Cash = 10 Cr

Calculate the following:

- EBITDA Margin
- Net Debt
- Net Debt/EBITDA

Show the calculations briefly and provide a short interpretatation of the Company's financial position
"""
```

```python
response=ask_gemini(prompt)
print(response)
```
