```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv("/content/fraud_transactions_synthetic.csv")
```

```python
import tensorflow as tf
from tensorflow import keras #one of the most important modules in tensorflow
from tensorflow.keras import layers #layers will help us in building the neural ntwork sequentially
```

```python
x=data.drop(['is_fraud'],axis=1) #all the input independent columns
y=data['is_fraud'] #output dependent column
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.80)
```

```python
x_train.shape[1] #these 10 neurons will be used in the input layer when we build our NN
```

```python
model=keras.Sequential([
    layers.Input(shape=(x_train.shape[1],),name='input_layer'),
    layers.Dense(16,activation='relu',name='hidden_layer_1'),#number of neurons in the 1st hidden layer - 16 neurons
    layers.Dense(32,activation='relu',name='hidden_layer_2'),#number of neurons in the 2nd hidden layer - 32 neurons
    layers.Dense(1,activation='sigmoid',name='output_layer') #number of neurons in the output layer -1 and sigmoid af since it is a case of binary classification
])
```

Task: Observe the difference in output accuracy basis the following:

- Increase the number of epochs from 10 to 20
- change the activation function from relu to tanh in both the hidden layers
- change the activation function from relu to tanh in 1st hidden layer and keep it same for the 2nd hidden layer
- change the activation function from sigmoid to - first relu then softmax then tanh in the output layer

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy', #binary_crossentropy is used for binary classification [sparse_categorical_crossentropy - for multiclass classification]
    metrics=['accuracy']
)
```

```python
model.fit(x_train,y_train,epochs=10)
```
